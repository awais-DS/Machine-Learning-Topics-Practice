```python
salary=50000 # My Salary

#My House Rent
House_rent=int(input("Enter Your House Rent : "))

# Gas Charges
gas_charges=float(input("Enter Gas Charges:"))

#Wifi Bills
Wifi_bill=float(input("Enter Wifi  bill:"))

#Total Expenses
total=House_rent+gas_charges+Wifi_bill
print(f"Your total Expenses are : {total}")

#Remaining Balance
remaining_expense=salary-total
print(f"Your Remaing Salary is {remaining_expense}")
```
