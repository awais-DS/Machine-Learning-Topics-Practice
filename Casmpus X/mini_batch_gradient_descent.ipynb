```python
import numpy as np
import random

class MBGDRegressor:
    def __init__(self, batch_size, learning_rate=0.01, epochs=50):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size

    def fit(self, x_train, y_train):
        # 1. Properly initialize parameters (intercept is scalar, coef is vector)
        self.intercept_ = 0.0
        self.coef_ = np.ones(x_train.shape[1])
        
        num_samples = x_train.shape[0]

        for i in range(self.epochs):
            for j in range(int(num_samples / self.batch_size)):
                # 2. Get random indices for the mini-batch
                idx = random.sample(range(num_samples), self.batch_size)
                
                # Extract batch data
                x_batch = x_train[idx]
                y_batch = y_train[idx]
                
                # 3. Calculate predictions for the batch
                y_hat = np.dot(x_batch, self.coef_) + self.intercept_
                
                # Calculate the error vector
                error = y_batch - y_hat

                # 4. Derivative with respect to intercept (bias)
                intercept_derivative = -2 * np.mean(error)
                self.intercept_ = self.intercept_ - (self.lr * intercept_derivative)

                # 5. Derivative with respect to coefficients (weights)
                # We use dot product of x_batch transposed and error, averaged by batch size
                coef_derivative = -2 * np.dot(x_batch.T, error) / self.batch_size
                self.coef_ = self.coef_ - (self.lr * coef_derivative)

        print("Training complete!")
        print("Intercept:", self.intercept_, "\nCoefficients:", self.coef_)

    def predict(self, x_test):
        return np.dot(x_test, self.coef_) + self.intercept_








        
```

```python
mbr=MBGDRegressor(batch_size=int(x_train.shape[0]/50),learning_rate=0.02,epochs=50)
```

```python
mbr.fit(x_train,y_train)
```

```python
y_pred= mbr.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```
