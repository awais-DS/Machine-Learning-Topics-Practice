In this Notebook I am going to do the two types of encoding 1.Ordinal Datra Encoding and 2.Label Data encoding.
Ordinal data Encoding:- When we have a a categoraical data in inputs columns which has order like(Low,Medium,High) called ordinal Data.So we transform the column from characters to string using ordinal encoding.
Label encoding:-When we have categorical Data in output feature so we use Label encoding and its just givv it representative values like(0 to A,1 to B,2 to C etc).
